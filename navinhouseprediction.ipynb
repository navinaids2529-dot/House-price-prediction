```python
import pandas as pd
```

```python
house=pd.read_csv('/content/House Price Prediction Dataset.csv')
house.head(20)
```

```python
import numpy as np
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
x = house[['Area', 'Bedrooms', 'Bathrooms', 'Floors', 'YearBuilt']]
y = house['Price']
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X_test['Area'],y_test,color='red',label='Actual Data')
plt.scatter(X_test['Area'],y_pred,color='blue',label='Regression Line')
plt.xlabel('Area')
plt.ylabel('Price')
plt.legend()
plt.title('House Price detection using LinearRegression')
plt.show()
```
